3. Следующим этапом преобразуем признак **"Опыт работы"**. Его текущий формат - это: **<Опыт работы: n лет m месяцев, периоды работы в различных компаниях…>**. 

Из столбца нам необходимо выделить общий опыт работы соискателя в месяцах, новый признак назовем "Опыт работы (месяц)"

Для начала обсудим условия решения задачи:
* Во-первых, в данном признаке есть пропуски. Условимся, что если мы встречаем пропуск, оставляем его как есть (функция-преобразование возвращает NaN)
* Во-вторых, в данном признаке есть скрытые пропуски. Для некоторых соискателей в столбце стоит значения "Не указано". Их тоже обозначим как NaN (функция-преобразование возвращает NaN)
* В-третьих, нас не интересует информация, которая описывается после указания опыта работы (периоды работы в различных компаниях)
* В-четвертых, у нас есть проблема: опыт работы может быть представлен только в годах или только месяцах. Например, можно встретить следующие варианты:
    * Опыт работы 3 года 2 месяца…
    * Опыт работы 4 года…
    * Опыт работы 11 месяцев…
    * Учитывайте эту особенность в вашем коде

Учитывайте эту особенность в вашем коде

В результате преобразования у вас должен получиться столбец, содержащий информацию о том, сколько месяцев проработал соискатель.
Выполните преобразование, ответьте на контрольные вопросы и удалите столбец **"Опыт работы"** из таблицы.


```python
import pandas as pd
import numpy as np
import re

df = pd.read_csv('hh_test.csv', sep=';')

def get_experience(arg):
    if pd.isna(arg) or arg == 'Не указано':
        return np.nan

    match = re.search(r'Опыт работы\s*(.*?)(?:,|$)', arg)

    if not match:
        return np.nan

    experience = match.group(1)

    years = re.search(r'(\d+)\s*(?:лет|года|год)', experience)
    months = re.search(r'(\d+)\s*месяц', experience)

    years_value = int(years.group(1)) if years else 0
    months_value = int(months.group(1)) if months else 0

    return years_value * 12 + months_value


df['Опыт работы (месяц)'] = df['Опыт работы'].apply(get_experience)

# Проверка результатов
print(df[['Опыт работы (месяц)']].head(10))
print(df['Опыт работы (месяц)'].describe())
print('Количество пропусков:', df['Опыт работы (месяц)'].isna().sum())

# Удаление исходного признака
df = df.drop(columns=['Опыт работы'])
```

Выполните преобразования и удалите столбец **"Город, переезд, командировки"** из таблицы.

```python
df = pd.read_csv('hh_test.csv', sep=';')


million_cities = [
    'Новосибирск', 'Екатеринбург', 'Нижний Новгород',
    'Казань', 'Челябинск', 'Омск', 'Самара',
    'Ростов-на-Дону', 'Уфа', 'Красноярск',
    'Пермь', 'Воронеж', 'Волгоград'
]


# Определение категории города
def get_city(arg):
    city = arg.split(' , ')[0]

    if city == 'Москва':
        return 'Москва'
    elif city == 'Санкт-Петербург':
        return 'Санкт-Петербург'
    elif city in million_cities:
        return 'город-миллионник'
    else:
        return 'другие'


# Определение готовности к переезду
def get_moving(arg):
    if 'не готов' in arg or 'не готова' in arg:
        return False
    elif 'готов к переезду' in arg or 'готова к переезду' in arg or 'хочу переехать' in arg:
        return True
    else:
        return False


# Определение готовности к командировкам
def get_business_trip(arg):
    if 'не готов к командировкам' in arg or 'не готова к командировкам' in arg:
        return False
    elif 'готов к командировкам' in arg or 'готова к командировкам' in arg:
        return True
    else:
        return False


# Создание новых признаков
df['Город'] = df['Город, переезд, командировки'].apply(get_city)
df['Готовность к переезду'] = df['Город, переезд, командировки'].apply(get_moving)
df['Готовность к командировкам'] = df['Город, переезд, командировки'].apply(get_business_trip)


# Проверка результатов
print(df['Город'].unique())
print(df['Готовность к переезду'].unique())
print(df['Готовность к командировкам'].unique())

print(df['Город'].value_counts())
print(df['Готовность к переезду'].value_counts())
print(df['Готовность к командировкам'].value_counts())


# Удаление исходного столбца
df = df.drop(columns=['Город, переезд, командировки'])
```

5. Рассмотрим поближе признаки **"Занятость"** и **"График"**. Сейчас признаки представляют собой набор категорий желаемой занятости (полная занятость, частичная занятость, проектная работа, волонтерство, стажировка) и желаемого графика работы (полный день, сменный график, гибкий график, удаленная работа, вахтовый метод).
На сайте hh.ru соискатель может указывать различные комбинации данных категорий, например:
* полная занятость, частичная занятость
* частичная занятость, проектная работа, волонтерство
* полный день, удаленная работа
* вахтовый метод, гибкий график, удаленная работа, полная занятость

Такой вариант признаков имеет множество различных комбинаций, а значит множество уникальных значений, что мешает анализу. Нужно это исправить!

Давайте создадим признаки-мигалки для каждой категории: если категория присутствует в списке желаемых соискателем, то в столбце на месте строки рассматриваемого соискателя ставится True, иначе - False.

Такой метод преобразования категориальных признаков называется One Hot Encoding и его схема представлена на рисунке ниже:
<img src=https://raw.githubusercontent.com/AndreyRysistov/DatasetsForPandas/main/ohe.jpg>
Выполните данное преобразование для признаков "Занятость" и "График", ответьте на контрольные вопросы, после чего удалите их из таблицы

```python
import pandas as pd
df = pd.read_csv('hh_test.csv', sep=';')

# Преобразуем признак «Занятость» методом One Hot Encoding
df = pd.concat([
    df,
    df['Занятость'].str.get_dummies(sep=', ').astype(bool)
], axis=1)

# Преобразуем признак «График» методом One Hot Encoding
df = pd.concat([
    df,
    df['График'].str.get_dummies(sep=', ').astype(bool)
], axis=1)

# Удаляем исходные столбцы
df.drop(columns=['Занятость', 'График'], inplace=True)

# Выводим первые 5 строк таблицы
print('Первые 5 строк таблицы:')
print(df.head())

# Выводим список новых признаков
print('\nНовые признаки:')
print(df.columns.tolist())

# Проверяем количество уникальных значений в новых столбцах
print('\nКоличество уникальных значений в новых признаках:')
print(df.select_dtypes(include='bool').nunique())

count = ((df['проектная работа'] == True) & 
         (df['волонтерство'] == True)).sum()

print('Количество людей (проектная работа+волонтерство):', count)
```

1. Перевести признак "Обновление резюме" из таблицы с резюме в формат datetime и достать из него дату. В тот же формат привести признак "date" из таблицы с валютами.
2. Выделить из столбца "ЗП" сумму желаемой заработной платы и наименование валюты, в которой она исчисляется. Наименование валюты перевести в стандарт ISO согласно с таблицей выше.
3. Присоединить к таблице с резюме таблицу с курсами по столбцам с датой и названием валюты (подумайте, какой тип объединения надо выбрать, чтобы в таблице с резюме сохранились данные о заработной плате, изначально представленной в рублях). Значение close для рубля заполнить единицей 1 (курс рубля самого к себе)
4. Умножить сумму желаемой заработной платы на присоединенный курс валюты (close) и разделить на пропорцию (обратите внимание на пропуски после объединения в этих столбцах), результат занести в новый столбец "ЗП (руб)".


```python

import pandas as pd

# 1. Загружаем таблицу курсов валют
exchange_rates = pd.read_csv('ExchangeRates.csv')

# 2. Преобразуем даты
exchange_rates['date'] = pd.to_datetime(
    exchange_rates['date'],
    format='%d/%m/%y'
)

df['Дата обновления'] = pd.to_datetime(
    df['Обновление резюме'],
    dayfirst=True,
    errors='coerce'
).dt.normalize()

# 3. Словарь соответствия валют
currency_mapping = {
    'руб.': 'RUB',
    'руб': 'RUB',
    'грн.': 'UAH',
    'грн': 'UAH',
    'USD': 'USD',
    'EUR': 'EUR',
    'бел.руб.': 'BYN',
    'KZT': 'KZT',
    'сум': 'UZS',
    'KGS': 'KGS',
    'AZN': 'AZN'
}

# 4. Извлекаем сумму и валюту
currency_pattern = (
    r'(бел\.руб\.|руб\.?|грн\.?|USD|EUR|KZT|KGS|AZN|сум)\s*$'
)

df['Валюта'] = df['ЗП'].str.extract(
    currency_pattern
)[0]

df['Сумма ЗП'] = (
    df['ЗП']
    .str.extract(r'(\d[\d\s]*)')[0]
    .str.replace(r'\s+', '', regex=True)
)

df['Сумма ЗП'] = pd.to_numeric(
    df['Сумма ЗП'],
    errors='coerce'
)

df['ISO'] = df['Валюта'].map(currency_mapping)

# 5. Проверяем распознавание зарплат и валют
if df['Сумма ЗП'].isna().any():
    raise ValueError('Есть зарплаты, которые не удалось распознать.')

if df['ISO'].isna().any():
    unknown = df.loc[df['ISO'].isna(), 'Валюта'].unique()
    raise ValueError(f'Неизвестные обозначения валют: {unknown}')

# 6. Подготавливаем таблицу курсов
rates = exchange_rates[
    ['currency', 'date', 'proportion', 'close']
].rename(columns={
    'currency': 'ISO',
    'date': 'Дата обновления',
    'proportion': 'Пропорция',
    'close': 'Курс'
})

# 7. Объединяем таблицы
df = df.merge(
    rates,
    on=['ISO', 'Дата обновления'],
    how='left',
    validate='many_to_one'
)

# 8. Для рублей курс равен единице
rub_mask = df['ISO'] == 'RUB'

df.loc[rub_mask, 'Курс'] = 1
df.loc[rub_mask, 'Пропорция'] = 1

# 9. Проверяем наличие курсов для иностранных валют
missing_rates = (
    df['ISO'].ne('RUB')
    & (df['Курс'].isna() | df['Пропорция'].isna())
)

if missing_rates.any():
    print('Примеры строк без курса:')
    print(df.loc[
        missing_rates,
        ['ЗП', 'ISO', 'Дата обновления']
    ].head(10))

    raise ValueError(
        f'Не найден курс для {missing_rates.sum()} резюме.'
    )

# 10. Рассчитываем зарплату в рублях
df['ЗП (руб)'] = (
    df['Сумма ЗП'] * df['Курс'] / df['Пропорция']
)

# 11. Проверяем результат
print('Первые 10 зарплат:')
print(df[['ЗП', 'ЗП (руб)']].head(10))

print('\nСредняя зарплата:', round(df['ЗП (руб)'].mean(), 2))
print('Медианная зарплата:', round(df['ЗП (руб)'].median(), 2))
print('Количество пропусков:', df['ЗП (руб)'].isna().sum())

# 12. Удаляем исходные и промежуточные столбцы
df.drop(
    columns=[
        'ЗП',
        'Дата обновления',
        'Сумма ЗП',
        'Валюта',
        'ISO',
        'Пропорция',
        'Курс'
    ],
    inplace=True
)

print('\nИтоговые столбцы:')
print(df.columns.tolist())
```

```python

import os
import pandas as pd
import numpy as np
import re
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# =====================================================
# 1. ЗАГРУЗКА ИСХОДНОГО ФАЙЛА
# =====================================================

file_path = 'hh_test.csv'

df = None

for encoding in ['utf-8-sig', 'cp1251']:
    for separator in [None, ';', ',', '\t']:
        try:
            temp_df = pd.read_csv(
                file_path,
                sep=separator,
                engine='python',
                encoding=encoding
            )

            if len(temp_df.columns) > 1:
                df = temp_df
                print('Файл успешно загружен!')
                print('Кодировка:', encoding)
                print(
                    'Разделитель:',
                    'автоматический' if separator is None else separator
                )
                break

        except Exception:
            continue

    if df is not None:
        break

if df is None:
    raise ValueError('Не удалось прочитать CSV-файл.')

df.columns = df.columns.astype(str).str.strip()

print('\nРазмер исходной таблицы:', df.shape)
print('Названия столбцов:', df.columns.tolist())


# =====================================================
# 2. ОБРАБОТКА ПРИЗНАКА «ОПЫТ РАБОТЫ»
# =====================================================

if 'Опыт работы' not in df.columns:
    raise KeyError('Столбец «Опыт работы» не найден.')


def get_experience(value):

    if pd.isna(value):
        return np.nan

    value = str(value).strip().lower()

    if value in ['', 'не указано']:
        return np.nan

    # Ищем годы
    years = re.search(
        r'(?<!\d)(\d{1,2})\s*(?:лет|года|год)(?![а-я])',
        value
    )

    # Ищем месяцы
    months = re.search(
        r'(?<!\d)(\d{1,2})\s*(?:месяцев|месяца|месяц)(?![а-я])',
        value
    )

    result = 0
    found = False

    if years:
        year_value = int(years.group(1))

        if 0 <= year_value <= 60:
            result += year_value * 12
            found = True

    if months:
        month_value = int(months.group(1))

        if 0 <= month_value <= 11:
            result += month_value
            found = True

    return result if found else np.nan


# Создаём числовой столбец
df['Опыт работы (месяц)'] = df['Опыт работы'].apply(
    get_experience
)

# Удаляем пропуски для статистического анализа
experience = df['Опыт работы (месяц)'].dropna()

if experience.empty:
    raise ValueError('Нет корректных значений опыта работы.')


# =====================================================
# 3. РАСЧЁТ СТАТИСТИЧЕСКИХ ПОКАЗАТЕЛЕЙ
# =====================================================

mode = experience.mode().tolist()
minimum = experience.min()
maximum = experience.max()
mean = experience.mean()
median = experience.median()

Q1 = experience.quantile(0.25)
Q3 = experience.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Определяем аномальные значения
anomalies = experience[
    (experience < lower_bound) |
    (experience > upper_bound)
]

# 99-й процентиль
display_max = experience.quantile(0.99)


# =====================================================
# 4. ПОСТРОЕНИЕ ГРАФИКОВ
# =====================================================

fig = make_subplots(
    rows=1,
    cols=2,
    subplot_titles=(
        'Распределение опыта работы',
        'Коробчатая диаграмма'
    ),
    horizontal_spacing=0.13
)

# Ограничиваем гистограмму 99-м процентилем
hist_data = experience[experience <= display_max]

# Гистограмма
fig.add_trace(
    go.Histogram(
        x=hist_data,
        xbins=dict(size=12),
        marker=dict(
            color='steelblue',
            line=dict(color='white', width=0.5)
        ),
        name='Опыт работы',
        hovertemplate=(
            'Опыт работы: %{x} месяцев<br>'
            'Количество: %{y}<extra></extra>'
        )
    ),
    row=1,
    col=1
)

# Коробчатая диаграмма
fig.add_trace(
    go.Box(
        x=experience,
        name='Опыт работы',
        marker=dict(color='darkorange'),
        line=dict(color='darkorange'),
        boxmean=True,
        boxpoints='outliers',
        hovertemplate=(
            'Опыт работы: %{x} месяцев<extra></extra>'
        )
    ),
    row=1,
    col=2
)

# Общие настройки
fig.update_layout(
    title=dict(
        text='Распределение признака «Опыт работы (месяц)»',
        x=0.5
    ),
    height=600,
    showlegend=False,
    template='plotly_white',
    bargap=0.08,
    font=dict(size=13)
)

fig.update_xaxes(
    title_text='Опыт работы (месяцы)',
    range=[0, display_max],
    dtick=12,
    showgrid=True,
    row=1,
    col=1
)

fig.update_xaxes(
    title_text='Опыт работы (месяцы)',
    range=[0, display_max],
    dtick=12,
    showgrid=True,
    row=1,
    col=2
)

fig.update_yaxes(
    title_text='Количество соискателей',
    rangemode='tozero',
    showgrid=True,
    row=1,
    col=1
)

# Создаём папку для графиков
os.makedirs('graphs', exist_ok=True)

# Сохраняем график
fig.write_html(
    'graphs/experience_distrib.html',
    include_plotlyjs=True
)

# Отображаем график
fig.show()


# =====================================================
# 5. ВЫВОД СТАТИСТИКИ
# =====================================================

print('\n' + '=' * 55)
print('СТАТИСТИЧЕСКИЙ АНАЛИЗ ОПЫТА РАБОТЫ')
print('=' * 55)

print('Всего строк в таблице:', len(df))
print('Количество заполненных значений:', len(experience))
print(
    'Количество пропусков:',
    df['Опыт работы (месяц)'].isna().sum()
)

print('\nОсновные показатели:')

print('Мода:', mode, 'месяцев')
print('Минимальный опыт:', minimum, 'месяцев')
print('Максимальный опыт:', maximum, 'месяцев')
print('Средний опыт:', round(mean, 2), 'месяцев')
print('Медиана:', median, 'месяцев')

print('\nКвартильный анализ:')

print('Первый квартиль (Q1):', Q1)
print('Третий квартиль (Q3):', Q3)
print('Межквартильный размах (IQR):', IQR)

print('Нижняя граница:', lower_bound)
print('Верхняя граница:', upper_bound)

print('\nАнализ аномальных значений:')

print('Количество аномалий:', len(anomalies))

anomaly_percent = len(anomalies) / len(experience) * 100

print('Доля аномалий:', round(anomaly_percent, 2), '%')

print(
    'Минимальная аномалия:',
    anomalies.min() if len(anomalies) else 'Нет'
)

print(
    'Максимальная аномалия:',
    anomalies.max() if len(anomalies) else 'Нет'
)

print('\n99-й процентиль:', display_max)

print('\n20 наибольших значений опыта работы:')

print(
    df.nlargest(20, 'Опыт работы (месяц)')[
        ['Опыт работы', 'Опыт работы (месяц)']
    ].to_string(index=False)
)

print('\nПервые 20 аномальных значений:')

print(sorted(anomalies.unique())[:20])

print('\n' + '=' * 55)
print('АНАЛИЗ ЗАВЕРШЕН')
print('График сохранён: graphs/experience_distrib.html')
print('=' * 55)
```

```python

import pandas as pd
import numpy as np
import re
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# =====================================================
# 1. ПРОВЕРКА ИСХОДНОГО ПРИЗНАКА
# =====================================================

salary_column = 'ЗП'

if salary_column not in df.columns:
    raise KeyError(
        f'Столбец {salary_column} не найден. '
        f'Доступные столбцы: {df.columns.tolist()}'
    )

# =====================================================
# 2. ПРЕОБРАЗОВАНИЕ ЗАРАБОТНОЙ ПЛАТЫ В ЧИСЛА
# =====================================================

def convert_salary(value):

    if pd.isna(value):
        return np.nan

    if isinstance(value, (int, float, np.integer, np.floating)):
        return float(value)

    value = str(value).strip()

    if value.lower() in ['', 'nan', 'не указано']:
        return np.nan

    # Убираем пробелы, в том числе неразрывные
    value = value.replace('\xa0', '').replace(' ', '')

    # Извлекаем числовое значение
    match = re.search(r'\d+(?:[.,]\d+)?', value)

    if match:
        return float(match.group().replace(',', '.'))

    return np.nan


# Создаем отдельный числовой столбец
df['ЗП числовая'] = df[salary_column].apply(
    convert_salary
)

# Получаем заполненные значения
salary = df['ЗП числовая'].dropna()

# Проверяем корректность данных
salary = salary[salary > 0]

if len(salary) == 0:
    raise ValueError('Не удалось получить корректные значения зарплаты.')

# =====================================================
# 3. РАСЧЕТ СТАТИСТИЧЕСКИХ ПОКАЗАТЕЛЕЙ
# =====================================================

minimum = salary.min()
maximum = salary.max()
mean = salary.mean()
median = salary.median()
mode = salary.mode()

Q1 = salary.quantile(0.25)
Q3 = salary.quantile(0.75)

IQR = Q3 - Q1

# Границы аномальных значений
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Поиск аномалий
anomalies = salary[
    (salary < lower_bound) |
    (salary > upper_bound)
]

# 99-й процентиль для отображения основной массы
display_max = salary.quantile(0.99)

# =====================================================
# 4. ПОСТРОЕНИЕ ДВУХ ДИАГРАММ
# =====================================================

fig = make_subplots(
    rows=1,
    cols=2,
    subplot_titles=(
        'Распределение желаемой зарплаты',
        'Коробчатая диаграмма'
    ),
    horizontal_spacing=0.13
)

# Гистограмма основной массы значений
hist_salary = salary[salary <= display_max]

fig.add_trace(
    go.Histogram(
        x=hist_salary,
        nbinsx=50,
        marker=dict(
            color='steelblue',
            line=dict(color='white', width=0.5)
        ),
        name='Заработная плата'
    ),
    row=1,
    col=1
)

# Коробчатая диаграмма по всем наблюдениям
fig.add_trace(
    go.Box(
        x=salary,
        name='Заработная плата',
        marker=dict(color='darkorange'),
        boxmean=True,
        boxpoints='outliers'
    ),
    row=1,
    col=2
)

# Настройка внешнего вида
fig.update_layout(
    title=dict(
        text='Распределение признака «ЗП»',
        x=0.5
    ),
    height=600,
    showlegend=False,
    template='plotly_white',
    bargap=0.08,
    font=dict(size=13)
)

# Настройка осей гистограммы
fig.update_xaxes(
    title_text='Желаемая заработная плата (руб.)',
    range=[0, display_max],
    tickformat=',.0f',
    nticks=12,
    tickangle=0,
    row=1,
    col=1
)

# Настройка осей коробчатой диаграммы
fig.update_xaxes(
    title_text='Желаемая заработная плата (руб.)',
    range=[0, display_max],
    tickformat=',.0f',
    nticks=12,
    tickangle=0,
    row=1,
    col=2
)

fig.update_yaxes(
    title_text='Количество соискателей',
    rangemode='tozero',
    row=1,
    col=1
)

fig.show()

# =====================================================
# 5. ВЫВОД РЕЗУЛЬТАТОВ
# =====================================================

print('\n' + '=' * 60)
print('АНАЛИЗ ПРИЗНАКА «ЗП»')
print('=' * 60)

print('\nОсновные показатели:')

print('Количество наблюдений:', len(salary))
print('Количество пропусков:',
      df['ЗП числовая'].isna().sum())

print('Минимальная зарплата:', round(minimum, 2), 'руб.')
print('Максимальная зарплата:', round(maximum, 2), 'руб.')
print('Средняя зарплата:', round(mean, 2), 'руб.')
print('Медианная зарплата:', round(median, 2), 'руб.')
print('Мода:', mode.tolist())

print('\nКвартильный анализ:')

print('Первый квартиль (Q1):', Q1)
print('Третий квартиль (Q3):', Q3)
print('Межквартильный размах (IQR):', IQR)

print('Нижняя граница:', lower_bound)
print('Верхняя граница:', upper_bound)

print('\nАнализ аномальных значений:')

print('Количество аномалий:', len(anomalies))

print(
    'Доля аномалий:',
    round(len(anomalies) / len(salary) * 100, 2),
    '%'
)

print('Минимальная аномалия:',
      anomalies.min() if len(anomalies) else 'Нет')

print('Максимальная аномалия:',
      anomalies.max() if len(anomalies) else 'Нет')

print('\nПроцентили:')

for p in [0.50, 0.75, 0.90, 0.95, 0.99]:
    print(
        f'{int(p * 100)}-й процентиль:',
        round(salary.quantile(p), 2),
        'руб.'
    )

print('\n20 самых больших значений зарплаты:')

print(
    df.nlargest(20, 'ЗП числовая')[
        [salary_column, 'ЗП числовая']
    ].to_string(index=False)
)

print('\n' + '=' * 60)
print('АНАЛИЗ ЗАВЕРШЕН')
print('=' * 60)
```

4. Постройте диаграмму, которая показывает зависимость **медианной** желаемой заработной платы (**"ЗП (руб)"**) от уровня образования (**"Образование"**). Используйте для диаграммы данные о резюме, где желаемая заработная плата меньше 1 млн рублей.
*Сделайте выводы по представленной диаграмме: для каких уровней образования наблюдаются наибольшие и наименьшие уровни желаемой заработной платы? Как вы считаете, важен ли признак уровня образования при прогнозировании заработной платы?*

```python

import os
import pandas as pd
import numpy as np
import re
import plotly.express as px

# =====================================================
# 1. ПОДГОТОВКА ДАННЫХ
# =====================================================

salary_column = 'ЗП'
education_column = 'Образование и ВУЗ'

# Проверяем наличие столбцов
if salary_column not in df.columns or education_column not in df.columns:
    raise KeyError(
        'Не найдены необходимые столбцы. '
        f'Доступные столбцы: {df.columns.tolist()}'
    )

data = df[[salary_column, education_column]].copy()


# =====================================================
# 2. ПРЕОБРАЗОВАНИЕ ЗАРАБОТНОЙ ПЛАТЫ
# =====================================================

def convert_salary(value):

    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    # Удаляем пробелы и неразрывные пробелы
    value = value.replace('\xa0', '').replace(' ', '')

    match = re.search(r'\d+(?:[.,]\d+)?', value)

    if match:
        return float(match.group().replace(',', '.'))

    return np.nan


data['ЗП (руб)'] = data[salary_column].apply(convert_salary)


# =====================================================
# 3. ВЫДЕЛЕНИЕ УРОВНЯ ОБРАЗОВАНИЯ
# =====================================================

def get_education(value):

    if pd.isna(value):
        return np.nan

    value = str(value).strip().lower()

    # Сначала проверяем более конкретные категории

    if 'неоконченное высшее' in value:
        return 'Неоконченное высшее'

    elif 'среднее специальное' in value:
        return 'Среднее специальное'

    elif 'высшее' in value:
        return 'Высшее'

    elif 'среднее образование' in value:
        return 'Среднее общее'

    return np.nan


data['Образование'] = data[education_column].apply(get_education)


# =====================================================
# 4. ФИЛЬТРАЦИЯ ДАННЫХ
# =====================================================

data = data[
    (data['ЗП (руб)'] > 0) &
    (data['ЗП (руб)'] < 1_000_000) &
    (data['Образование'].notna())
].copy()

if data.empty:
    raise ValueError('После фильтрации не осталось данных.')


# =====================================================
# 5. РАСЧЕТ МЕДИАННОЙ ЗАРАБОТНОЙ ПЛАТЫ
# =====================================================

result = (
    data.groupby('Образование')['ЗП (руб)']
    .agg(['median', 'count'])
    .reset_index()
    .rename(columns={
        'median': 'Медианная зарплата',
        'count': 'Количество резюме'
    })
    .sort_values(
        'Медианная зарплата',
        ascending=False
    )
)

print('\nМедианная зарплата по уровню образования:')
print(result.to_string(index=False))


# =====================================================
# 6. ПОСТРОЕНИЕ ДИАГРАММЫ
# =====================================================

fig = px.bar(
    result,
    x='Образование',
    y='Медианная зарплата',
    text='Медианная зарплата',
    title='Зависимость медианной желаемой зарплаты от уровня образования',
    hover_data=['Количество резюме'],
    color='Образование',
    labels={
        'Образование': 'Уровень образования',
        'Медианная зарплата': 'Медианная зарплата (руб.)'
    }
)

fig.update_traces(
    texttemplate='%{text:,.0f} ₽',
    textposition='outside'
)

fig.update_layout(
    template='plotly_white',
    height=550,
    width=1000,
    showlegend=False,
    title_x=0.5,
    xaxis_title='Уровень образования',
    yaxis_title='Медианная желаемая зарплата (руб.)'
)

fig.update_yaxes(
    tickformat=',.0f',
    rangemode='tozero'
)

# Создаём папку для графиков
os.makedirs('graphs', exist_ok=True)

# Сохраняем график
fig.write_html(
    'graphs/education_salary.html',
    include_plotlyjs=True
)

# Отображаем график
fig.show()


# =====================================================
# 7. ВЫВОД РЕЗУЛЬТАТОВ
# =====================================================

highest = result.loc[
    result['Медианная зарплата'].idxmax()
]

lowest = result.loc[
    result['Медианная зарплата'].idxmin()
]

print('\nКоличество резюме после фильтрации:', len(data))

print('\nНаибольшая медианная зарплата:')

print(
    highest['Образование'],
    '-',
    round(highest['Медианная зарплата']),
    'руб.'
)

print('\nНаименьшая медианная зарплата:')

print(
    lowest['Образование'],
    '-',
    round(lowest['Медианная зарплата']),
    'руб.'
)

print('\nГрафик сохранён: graphs/education_salary.html')
```

5. Постройте диаграмму, которая показывает распределение желаемой заработной платы (**"ЗП (руб)"**) в зависимости от города (**"Город"**). Используйте для диаграммы данные о резюме, где желая заработная плата меньше 1 млн рублей.
*Сделайте выводы по полученной диаграмме: как соотносятся медианные уровни желаемой заработной платы и их размах в городах? Как вы считаете, важен ли признак города при прогнозировании заработной платы?*

```python
import os
import pandas as pd
import numpy as np
import re
import plotly.express as px

# =====================================================
# 1. ЗАГРУЗКА ФАЙЛА
# =====================================================

file_path = 'hh_test.csv'

encodings = ['utf-8-sig', 'utf-8', 'cp1251']
separators = [';', ',', '\t']

df = None
last_error = None

for encoding in encodings:
    for separator in separators:
        try:
            temp_df = pd.read_csv(
                file_path,
                sep=separator,
                encoding=encoding,
                encoding_errors='replace',
                engine='python',
                on_bad_lines='warn',
                quotechar='"'
            )

            if len(temp_df.columns) > 1 and len(temp_df) > 0:
                df = temp_df
                print('Файл успешно загружен!')
                print('Кодировка:', encoding)
                print('Разделитель:', repr(separator))
                break

        except Exception as error:
            last_error = error

    if df is not None:
        break

if df is None:
    raise ValueError(
        f'Не удалось загрузить файл. Ошибка: {last_error}'
    )

df.columns = df.columns.astype(str).str.strip()

print('\nРазмер исходной таблицы:', df.shape)
print('\nНазвания столбцов:')
print(df.columns.tolist())


# =====================================================
# 2. ПРОВЕРКА СТОЛБЦОВ
# =====================================================

salary_column = 'ЗП'
city_column = 'Город, переезд, командировки'

for column in [salary_column, city_column]:
    if column not in df.columns:
        raise KeyError(
            f'Не найден столбец "{column}". '
            f'Доступные столбцы: {df.columns.tolist()}'
        )


# =====================================================
# 3. ПРЕОБРАЗОВАНИЕ ЗАРАБОТНОЙ ПЛАТЫ
# =====================================================

def convert_salary(value):

    if pd.isna(value):
        return np.nan

    if isinstance(value, (int, float, np.number)):
        return float(value)

    value = str(value).strip()
    value = value.replace('\xa0', '').replace(' ', '')

    match = re.search(r'\d+(?:[.,]\d+)?', value)

    if match:
        return float(match.group().replace(',', '.'))

    return np.nan


df['ЗП (руб)'] = df[salary_column].apply(convert_salary)


# =====================================================
# 4. ИЗВЛЕЧЕНИЕ ГОРОДА
# =====================================================

def get_city(value):

    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    if not value:
        return np.nan

    city = value.split(',')[0].strip()

    if city.lower() in ['', 'не указано', 'nan']:
        return np.nan

    return city


df['Город'] = df[city_column].apply(get_city)


# =====================================================
# 5. ФИЛЬТРАЦИЯ ДАННЫХ
# =====================================================

filtered_df = df[
    (df['ЗП (руб)'] > 0) &
    (df['ЗП (руб)'] < 1_000_000) &
    (df['Город'].notna())
].copy()

if filtered_df.empty:
    raise ValueError(
        'После фильтрации не осталось данных. '
        'Проверь исходные значения зарплаты и города.'
    )


# =====================================================
# 6. РАСЧЁТ СТАТИСТИКИ ПО ГОРОДАМ
# =====================================================

city_stats = (
    filtered_df
    .groupby('Город')['ЗП (руб)']
    .agg(
        Медиана='median',
        Q1=lambda x: x.quantile(0.25),
        Q3=lambda x: x.quantile(0.75),
        Количество='count'
    )
    .reset_index()
)

city_stats['Межквартильный размах'] = (
    city_stats['Q3'] - city_stats['Q1']
)

# Оставляем города, где не менее 10 резюме
popular_cities = city_stats[
    city_stats['Количество'] >= 10
].copy()

if popular_cities.empty:
    raise ValueError(
        'Нет городов, в которых набралось хотя бы 10 резюме.'
    )


# =====================================================
# 7. ПОДГОТОВКА ДАННЫХ ДЛЯ ДИАГРАММЫ
# =====================================================

plot_df = filtered_df[
    filtered_df['Город'].isin(popular_cities['Город'])
].copy()

city_order = (
    popular_cities
    .sort_values('Медиана')['Город']
    .tolist()
)


# =====================================================
# 8. ПОСТРОЕНИЕ КОРОБЧАТОЙ ДИАГРАММЫ
# =====================================================

fig = px.box(
    plot_df,
    x='ЗП (руб)',
    y='Город',
    category_orders={'Город': city_order},
    points='outliers',
    title='Распределение желаемой заработной платы по городам',
    hover_data=['ЗП (руб)'],
    labels={
        'ЗП (руб)': 'Желаемая заработная плата, руб.',
        'Город': 'Город'
    }
)

fig.update_layout(
    template='plotly_white',
    title_x=0.5,
    height=max(600, len(city_order) * 35),
    width=1100,
    showlegend=False,
    font=dict(size=12)
)

fig.update_xaxes(
    tickformat=',.0f',
    rangemode='tozero'
)

# =====================================================
# 9. СОХРАНЕНИЕ ГРАФИКА
# =====================================================

os.makedirs('graphs', exist_ok=True)

fig.write_html(
    'graphs/salary_by_city.html',
    include_plotlyjs=True
)

fig.show()

print('\nГрафик сохранён: graphs/salary_by_city.html')


# =====================================================
# 10. ВЫВОД РЕЗУЛЬТАТОВ
# =====================================================

print('\n' + '=' * 60)
print('АНАЛИЗ ЗАРАБОТНОЙ ПЛАТЫ ПО ГОРОДАМ')
print('=' * 60)

print('Всего резюме:', len(df))
print('После фильтрации:', len(filtered_df))
print('Количество городов на диаграмме:', len(popular_cities))

print('\nСтатистика по городам:')

print(
    popular_cities
    .sort_values('Медиана', ascending=False)
    .round(2)
    .to_string(index=False)
)


# Город с максимальной медианной зарплатой
highest = popular_cities.loc[
    popular_cities['Медиана'].idxmax()
]

# Город с минимальной медианной зарплатой
lowest = popular_cities.loc[
    popular_cities['Медиана'].idxmin()
]

# Город с наибольшим разбросом зарплат
widest = popular_cities.loc[
    popular_cities['Межквартильный размах'].idxmax()
]

# Город с наименьшим разбросом зарплат
narrowest = popular_cities.loc[
    popular_cities['Межквартильный размах'].idxmin()
]

print('\nНаибольшая медианная зарплата:')
print(
    highest['Город'],
    '-',
    round(highest['Медиана']),
    'руб.'
)

print('\nНаименьшая медианная зарплата:')
print(
    lowest['Город'],
    '-',
    round(lowest['Медиана']),
    'руб.'
)

print('\nНаибольший разброс зарплат:')
print(
    widest['Город'],
    '-',
    round(widest['Межквартильный размах']),
    'руб.'
)

print('\nНаименьший разброс зарплат:')
print(
    narrowest['Город'],
    '-',
    round(narrowest['Межквартильный размах']),
    'руб.'
)

print('\n' + '=' * 60)
print('АНАЛИЗ ЗАВЕРШЕН')
print('=' * 60)
```

6. Постройте **многоуровневую столбчатую диаграмму**, которая показывает зависимость медианной заработной платы (**"ЗП (руб)"**) от признаков **"Готовность к переезду"** и **"Готовность к командировкам"**. Проанализируйте график, сравнив уровень заработной платы в категориях.

```python
import pandas as pd
import numpy as np
import re
import os
import plotly.express as px

# Проверяем наличие необходимых столбцов
required_columns = ['ЗП (руб)', 'Город, переезд, командировки']

for column in required_columns:
    if column not in df.columns:
        raise KeyError(
            f'Столбец "{column}" не найден. '
            f'Доступные столбцы: {df.columns.tolist()}'
        )

# Функция для извлечения готовности к переезду и командировкам
def get_mobility_info(value):
    if pd.isna(value):
        return pd.Series([None, None])

    value = str(value).lower()

    # Определяем готовность к переезду
    if 'не готов к переезду' in value or 'не готова к переезду' in value:
        relocation = 'Не готов'
    elif 'готов к переезду' in value or 'готова к переезду' in value:
        relocation = 'Готов'
    else:
        relocation = 'Не указано'

    # Определяем готовность к командировкам
    if 'не готов к командировкам' in value or 'не готова к командировкам' in value:
        business_trips = 'Не готов'
    elif 'готов к командировкам' in value or 'готова к командировкам' in value:
        business_trips = 'Готов'
    else:
        business_trips = 'Не указано'

    return pd.Series([relocation, business_trips])


# Создаем два новых столбца
df[['Готовность к переезду', 'Готовность к командировкам']] = (
    df['Город, переезд, командировки'].apply(get_mobility_info)
)

# Преобразуем зарплату в числовой формат
df['ЗП (руб)'] = pd.to_numeric(df['ЗП (руб)'], errors='coerce')

# Удаляем строки с отсутствующими значениями
filtered_df = df.dropna(
    subset=[
        'ЗП (руб)',
        'Готовность к переезду',
        'Готовность к командировкам'
    ]
).copy()

# Убираем некорректные значения зарплаты
filtered_df = filtered_df[
    (filtered_df['ЗП (руб)'] > 0) &
    (filtered_df['ЗП (руб)'] < 1_000_000)
]

# Группируем данные по двум признакам и рассчитываем медианную зарплату
salary_stats = (
    filtered_df
    .groupby(
        ['Готовность к переезду', 'Готовность к командировкам'],
        observed=True
    )['ЗП (руб)']
    .agg(['median', 'count'])
    .reset_index()
)

salary_stats = salary_stats.rename(
    columns={
        'median': 'Медианная зарплата',
        'count': 'Количество резюме'
    }
)

# Проверяем результат группировки
print('Таблица медианной зарплаты:')
print(salary_stats.to_string(index=False))

# Создаем многоуровневую столбчатую диаграмму
fig = px.bar(
    salary_stats,
    x='Готовность к переезду',
    y='Медианная зарплата',
    color='Готовность к командировкам',
    barmode='group',
    text_auto='.0f',
    hover_data=['Количество резюме'],
    title='Зависимость медианной заработной платы от готовности к переезду и командировкам',
    labels={
        'Готовность к переезду': 'Готовность к переезду',
        'Медианная зарплата': 'Медианная зарплата, руб.',
        'Готовность к командировкам': 'Готовность к командировкам'
    }
)

fig.update_layout(
    xaxis_title='Готовность к переезду',
    yaxis_title='Медианная зарплата, руб.',
    legend_title='Готовность к командировкам',
    template='plotly_white',
    height=600
)

fig.update_traces(
    textposition='outside'
)

# Создаем папку для графиков, если ее нет
os.makedirs('graphs', exist_ok=True)

# Сохраняем график
fig.write_html('graphs/salary_mobility.html')

# Отображаем график
fig.show()
```

```python
import pandas as pd
import numpy as np
import re
import os
import plotly.express as px

# Проверяем наличие исходных столбцов
required_columns = [
    'ЗП (руб)',
    'Пол, возраст',
    'Образование и ВУЗ'
]

for column in required_columns:
    if column not in df.columns:
        raise KeyError(
            f'Столбец "{column}" не найден. '
            f'Доступные столбцы: {df.columns.tolist()}'
        )

# 1. Извлекаем возраст из столбца "Пол, возраст"
df['Возраст'] = pd.to_numeric(
    df['Пол, возраст'].astype(str).str.extract(r'(\d+)', expand=False),
    errors='coerce'
)

# 2. Извлекаем уровень образования
def get_education(value):
    if pd.isna(value):
        return None

    value = str(value).lower()

    if 'неоконченное высшее' in value:
        return 'Неоконченное высшее'
    elif 'среднее специальное' in value:
        return 'Среднее специальное'
    elif 'высшее' in value:
        return 'Высшее'
    elif 'среднее образование' in value:
        return 'Среднее общее'
    else:
        return None

df['Образование'] = df['Образование и ВУЗ'].apply(get_education)

# 3. Преобразуем зарплату и возраст в числовой формат
df['ЗП (руб)'] = pd.to_numeric(df['ЗП (руб)'], errors='coerce')
df['Возраст'] = pd.to_numeric(df['Возраст'], errors='coerce')

# 4. Удаляем строки с пропущенными и некорректными значениями
filtered_df = df.dropna(
    subset=['ЗП (руб)', 'Возраст', 'Образование']
).copy()

filtered_df = filtered_df[
    (filtered_df['ЗП (руб)'] > 0) &
    (filtered_df['ЗП (руб)'] < 1_000_000) &
    (filtered_df['Возраст'] >= 14) &
    (filtered_df['Возраст'] <= 100)
]

filtered_df['Возраст'] = filtered_df['Возраст'].astype(int)

# 5. Строим сводную таблицу
pivot_table = pd.pivot_table(
    filtered_df,
    values='ЗП (руб)',
    index='Возраст',
    columns='Образование',
    aggfunc='median'
)

pivot_table = pivot_table.sort_index()

# 6. Выводим сводную таблицу
print('Сводная таблица медианной заработной платы:')
print(pivot_table.round(0).to_string())

# Строим тепловую карту с перевернутыми осями
fig = px.imshow(
    pivot_table.T,
    labels={
        'x': 'Возраст',
        'y': 'Образование',
        'color': 'Медианная зарплата, руб.'
    },
    title='Зависимость медианной заработной платы от возраста и образования',
    text_auto='.0f',
    aspect='auto',
    color_continuous_scale='YlOrRd'
)

fig.update_layout(
    xaxis_title='Возраст',
    yaxis_title='Образование',
    height=600,
    template='plotly_white'
)

# Сохраняем график
os.makedirs('graphs', exist_ok=True)
fig.write_html('graphs/salary_age_education.html')

# Отображаем график
fig.show()

```

8. Постройте **диаграмму рассеяния**, показывающую зависимость опыта работы (**"Опыт работы (месяц)"**) от возраста (**"Возраст"**). Опыт работы переведите из месяцев в года, чтобы признаки были в едином масштабе. Постройте на графике дополнительно прямую, проходящую через точки (0, 0) и (100, 100). Данная прямая соответствует значениям, когда опыт работы равен возрасту человека. Точки, лежащие на этой прямой и выше нее - аномалии в наших данных (опыт работы больше либо равен возрасту соискателя)

```python
import pandas as pd
import numpy as np
import re
import os
import plotly.graph_objects as go

# 1. Проверяем наличие исходных столбцов
required_columns = ['Пол, возраст', 'Опыт работы']

for column in required_columns:
    if column not in df.columns:
        raise KeyError(
            f'Столбец "{column}" не найден. '
            f'Доступные столбцы: {df.columns.tolist()}'
        )

# 2. Извлекаем возраст, если отдельного столбца еще нет
if 'Возраст' not in df.columns:
    df['Возраст'] = pd.to_numeric(
        df['Пол, возраст'].astype(str).str.extract(
            r'(\d+)', expand=False
        ),
        errors='coerce'
    )

# 3. Функция перевода опыта работы в месяцы
def convert_experience(value):
    if pd.isna(value):
        return np.nan

    value = str(value).lower()

    years = re.search(r'(\d+)\s*(?:лет|года|год)', value)
    months = re.search(r'(\d+)\s*(?:месяцев|месяца|месяц)', value)

    total_months = 0

    if years:
        total_months += int(years.group(1)) * 12

    if months:
        total_months += int(months.group(1))

    return total_months

# 4. Создаем столбец опыта работы в месяцах,
# если он еще не существует
if 'Опыт работы (месяц)' not in df.columns:
    df['Опыт работы (месяц)'] = df['Опыт работы'].apply(
        convert_experience
    )

# Преобразуем данные в числовой формат
df['Возраст'] = pd.to_numeric(df['Возраст'], errors='coerce')
df['Опыт работы (месяц)'] = pd.to_numeric(
    df['Опыт работы (месяц)'],
    errors='coerce'
)

# 5. Переводим опыт работы из месяцев в годы
df['Опыт работы (год)'] = df['Опыт работы (месяц)'] / 12

# 6. Удаляем строки с отсутствующими и некорректными значениями
filtered_df = df.dropna(
    subset=['Возраст', 'Опыт работы (год)']
).copy()

filtered_df = filtered_df[
    (filtered_df['Возраст'] >= 14) &
    (filtered_df['Возраст'] <= 100) &
    (filtered_df['Опыт работы (год)'] >= 0)
]

# 7. Определяем аномалии:
# опыт работы больше или равен возрасту
filtered_df['Аномалия'] = np.where(
    filtered_df['Опыт работы (год)'] >= filtered_df['Возраст'],
    'Аномалия',
    'Нормальные значения'
)

# 8. Выводим статистику
anomalies = filtered_df[
    filtered_df['Аномалия'] == 'Аномалия'
]

print('Количество резюме:', len(filtered_df))
print('Количество аномалий:', len(anomalies))
print(
    'Доля аномалий:',
    round(len(anomalies) / len(filtered_df) * 100, 2),
    '%'
)

# 9. Строим диаграмму рассеяния
fig = go.Figure()

# Нормальные значения
normal = filtered_df[
    filtered_df['Аномалия'] == 'Нормальные значения'
]

fig.add_trace(go.Scattergl(
    x=normal['Возраст'],
    y=normal['Опыт работы (год)'],
    mode='markers',
    name='Нормальные значения',
    marker=dict(
        color='royalblue',
        size=5,
        opacity=0.5
    )
))

# Аномальные значения
fig.add_trace(go.Scattergl(
    x=anomalies['Возраст'],
    y=anomalies['Опыт работы (год)'],
    mode='markers',
    name='Аномалии',
    marker=dict(
        color='red',
        size=7,
        opacity=0.8
    )
))

# 10. Добавляем прямую от (0, 0) до (100, 100)
fig.add_trace(go.Scatter(
    x=[0, 100],
    y=[0, 100],
    mode='lines',
    name='Опыт работы равен возрасту',
    line=dict(
        color='black',
        width=2,
        dash='dash'
    )
))

# 11. Настраиваем график
fig.update_layout(
    title='Зависимость опыта работы от возраста соискателей',
    xaxis_title='Возраст, лет',
    yaxis_title='Опыт работы, лет',
    template='plotly_white',
    height=700,
    legend_title='Категории'
)

# Одинаковый масштаб осей для правильного отображения диагонали
fig.update_xaxes(range=[0, 100])
fig.update_yaxes(range=[0, 100], scaleanchor='x', scaleratio=1)

# 12. Сохраняем график
os.makedirs('graphs', exist_ok=True)
fig.write_html('graphs/experience_age_scatter.html')

# 13. Отображаем график
fig.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Переводим возраст в логарифмический масштаб
log_age = np.log(df['Возраст'])

# 2. Вычисляем среднее и стандартное отклонение
mean_log_age = log_age.mean()
std_log_age = log_age.std()

# 3. Строим распределение возраста
plt.figure(figsize=(12, 6))
histplot = sns.histplot(log_age, bins=30)

# Добавляем линию среднего
histplot.axvline(mean_log_age, color='black', lw=2,
                 label='Среднее')

# Добавляем границы трех сигм
histplot.axvline(mean_log_age - 3 * std_log_age,
                 color='red', linestyle='--', lw=2,
                 label='-3 сигмы')

histplot.axvline(mean_log_age + 3 * std_log_age,
                 color='red', linestyle='--', lw=2,
                 label='+3 сигмы')

plt.title('Распределение возраста в логарифмическом масштабе')
plt.xlabel('Логарифм возраста')
plt.ylabel('Количество резюме')
plt.legend()
plt.grid(alpha=0.2)
plt.show()

# 4. Определяем асимметрию распределения
skewness = log_age.skew()

print('Асимметрия распределения:', round(skewness, 3))

if skewness > 0:
    print('Распределение имеет правостороннюю асимметрию.')
elif skewness < 0:
    print('Распределение имеет левостороннюю асимметрию.')
else:
    print('Распределение симметрично.')

# 5. Находим выбросы методом z-отклонений
z_scores = (log_age - mean_log_age) / std_log_age

# Используем 4 сигмы в правую сторону
outliers = z_scores > 4

# 6. Выводим таблицу найденных выбросов
outliers_table = df.loc[outliers].copy()

print('\nКоличество найденных выбросов:', len(outliers_table))
print('\nТаблица резюме с выбросами:')

display(outliers_table)

# 7. Определяем возраст соискателей, попавших в выбросы
if not outliers_table.empty:
    print('\nМинимальный возраст выбросов:',
          outliers_table['Возраст'].min())

    print('Максимальный возраст выбросов:',
          outliers_table['Возраст'].max())

    print('\nРаспределение возраста среди выбросов:')
    display(outliers_table['Возраст'].value_counts().sort_index())
else:
    print('Выбросы не обнаружены.')

# 8. Удаляем выбросы из исходной таблицы
df = df.loc[~outliers].copy()

# 9. Проверяем результат очистки
print('\nРезультаты очистки:')
print('Количество строк после удаления выбросов:', df.shape[0])
print('Количество оставшихся выбросов:',
      ((np.log(df['Возраст']) - mean_log_age) / std_log_age > 4).sum())
```
